# Daily Screen Time and Calorie Intake – Phase 2 (EDA & Hypothesis Testing)
This notebook performs exploratory data analysis and hypothesis testing for the dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
plt.rcParams['figure.figsize'] = (6,4)
```

```python
# Load dataset
df = pd.read_csv('data/daily_screen_calories.csv')
df.head()
```

```python
# Inspect dataset
df.info()
df.describe()
df.isna().sum()
```

```python
# Convert date column
df['Date'] = pd.to_datetime(df['Date'])
df.sort_values('Date', inplace=True)
df.head()
```

```python
# Histogram - Screen Time
sns.histplot(df['Screen_Time'], kde=True)
plt.title('Daily Screen Time Distribution (minutes)')
plt.xlabel('Screen Time (min)')
plt.show()
```

```python
# Histogram - Total Calories
sns.histplot(df['Total_Calories'], kde=True)
plt.title('Daily Calorie Intake Distribution (kcal)')
plt.xlabel('Calories (kcal)')
plt.show()
```

```python
# Scatter plot
plt.scatter(df['Screen_Time'], df['Total_Calories'])
plt.title('Screen Time vs Calorie Intake')
plt.xlabel('Screen Time (min)')
plt.ylabel('Total Calories (kcal)')
plt.show()
```

```python
# Pearson Correlation Test
corr, p_value = pearsonr(df['Screen_Time'], df['Total_Calories'])
corr, p_value
```

```python
# 📌 Phase 2 Findings (Automatically Generated)

import numpy as np

print("=== Phase 2 Findings ===\n")

num_days = len(df)
mean_screen = df['Screen_Time'].mean()
mean_calories = df['Total_Calories'].mean()

corr, p_value = pearsonr(df['Screen_Time'], df['Total_Calories'])

print(f"Number of days: {num_days}")
print(f"Mean Screen Time: {mean_screen:.2f} minutes")
print(f"Mean Total Calories: {mean_calories:.2f} kcal")
print(f"Correlation Coefficient (r): {corr:.4f}")
print(f"P-value: {p_value:.6f}")

# Interpretation
if p_value < 0.05:
    significance = "statistically significant"
else:
    significance = "NOT statistically significant"

print("\nInterpretation:")
print(f"- There is a strong positive correlation between screen time and calorie intake (r = {corr:.2f}).")
print(f"- The p-value ({p_value:.2e}) indicates that this relationship is {significance}.")
print("- This suggests that as screen time increases, daily calorie intake tends to increase as well.")
```
